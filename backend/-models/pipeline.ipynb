```python
from pathlib import Path
import shutil
import logging
from datetime import datetime

from FHBmodels.spikeExtract import run_spike_pipeline
from FHBmodels.spikeClassification import move_good_oriented_spikes
from FHBmodels.fhbScore import fhb_detect

intermediate_folder_name = "sdsu_" + datetime.now().strftime("%Y%m%d_%H%M%S")
RESULTS_ROOT = Path("./results/spikeDetect")

SPIKE_YOLO_MODEL_PATH = "./FHBmodels/spikeDetection/best_spike_YOLO11_freeze0_RS2.pt"
SPIKE_CLASSIFIER_PATH = "./FHBmodels/orientationClassification/small_3dcnn_best.pth"
FHB_MODEL_PATH = "./FHBmodels/FHBdetection/best_FHB_field_trainKSU_fineTune.pt"
```

```python
run_spike_pipeline(
    images_dir="./data/",
    model_path=SPIKE_YOLO_MODEL_PATH,
    max_crops=50,
    buffer_px=30,
    yolo_run_name=intermediate_folder_name,
    results_root=RESULTS_ROOT
)
```

```python
move_good_oriented_spikes(
    run_name=intermediate_folder_name,
    model_path=SPIKE_CLASSIFIER_PATH,
    results_root=RESULTS_ROOT,
    input_subdir="SpikeletCrops_30px",
    output_subdir="SpikeletCrops_30px_good",
)
```
